sampling distribution

````python
sample_mean = []
````

````python
for i in range (1000):

  sample = np.random.choice(
      nilai,
      size = 2,
      replace = False
  )

  mean = np.mean (sample)

  sample_mean.append(mean)
````

Visualisasi Menggunakan Matplotlib

````python
import matplotlib.pyplot as plt
````

````python
#Line Plot
x = [1,2,3,4,5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y)
plt.show()
````

````python
#Bar Plot

Kategori = ["A", "B","C", "D", "E"]
Nilai = [90, 30, 30, 40, 70]

plt.bar(Kategori, Nilai)
plt.show()
````

````python
#Histogram

nilai = [60, 70, 80, 90, 100]


plt.hist(nilai)
plt.show()
````

````python
populasi = np.random.normal(
    loc = 70, #mean
    scale = 5, #std
    size = 10000 #quantity data
)
````

````python
plt.hist(populasi, bins=50)
plt.title("distibusi populasi")
plt.xlabel("nilai")
plt.ylabel("frekuensi")
plt.show()
````

Scipy

````python
from scipy import stats
````

````python
distribusi_normal = stats.norm.rvs (
    loc = 70,
    scale = 5,
    size = 10000
)

plt.hist(distribusi_normal, bins=50)
plt.title("distibusi populasi")
plt.xlabel("nilai")
plt.ylabel("frekuensi")
plt.show()

````

````python
distribusi_exponensial = stats.expon.rvs (
    loc = 70,
    size = 10000
)

plt.hist(distribusi_exponensial , bins=50)
plt.title("distibusi populasi")
plt.xlabel("nilai")
plt.ylabel("frekuensi")
plt.show()
````

````python
distribusi_poisson = stats.poisson.rvs (
    mu=5, #rata rata
    size=10000 #Quantity data
)

plt.hist(distribusi_poisson , bins=50)
plt.title("distibusi populasi")
plt.xlabel("nilai")
plt.ylabel("frekuensi")
plt.show()
````
